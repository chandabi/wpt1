```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df=pd.read_csv('Boston (3).csv')
from sklearn.linear_model import LinearRegression
from sklearn import linear_model
from sklearn.metrics import mean_squared_error,mean_absolute_error
print(df)
print(df.info())
print(df.head())
plt.scatter(df.rm,df.rad,color='red',marker='*')
plt.xlabel('Built_up_area')
plt.ylabel('Rent')
plt.title('scatterplot')
plt.show()
x=df.drop('rad',axis=1)
print(x)
y=df.rad
print(y)
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.33,random_state=1)
print("xtrainshape:",xtrain.shape)
print("xtestshape:",xtest.shape)
print("ytrainshape:",ytrain.shape)
print("ytestshape:",ytest.shape)
reg=linear_model.LinearRegression()
reg.fit(x,y)
y_pred=reg.predict(xtest)
print(reg.coef_)
print(reg.intercept_)
z=9.7900215*650+1128.879253617697
print(z)
train_score=reg.score(xtrain,ytrain)
test_score=reg.score(xtest,ytest)
print('train score(x-squarred):',train_score)
print('test score(r-squarred):',test_score)
mse=mean_squared_error(ytest,y_pred)
print('mean squarred error:',mse)
mae=mean_absolute_error(ytest,y_pred)
print('mean absolute error:',mae)
```
